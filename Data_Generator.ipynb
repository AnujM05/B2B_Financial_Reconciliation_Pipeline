```python
import pandas as pd
import random
from datetime import datetime, timedelta

# 1. Base Setup: Generating the clean core data
random.seed(42) # Ensures you get the exact same messy dataset every time you run it
base_date = datetime(2026, 9, 1)

# Realistic corporate entities common in Indian B2B transactions
customers = [
    "Reliance Retail", "HDFC Bank Ltd", "Tata Consultancy Services", 
    "Infosys Limited", "Wipro Tech", "Zomato Media", 
    "Swiggy Delivery", "Flipkart Online", "Amazon India", "Paytm Digital"
]

transactions = []
for i in range(1, 101):
    transactions.append({
        "Order_ID": f"TXN-{1000+i}",
        "Transaction_Date": base_date + timedelta(days=random.randint(0, 30)),
        "Customer_Name": random.choice(customers),
        "Amount_Billed": round(random.uniform(5000, 50000), 2)
    })

# Create the Internal Sales Ledger (Dataset A)
ledger_df = pd.DataFrame(transactions)
ledger_df.to_csv("internal_sales_ledger.csv", index=False)

# 2. Bank Statement Setup: Duplicating and mutating the data
bank_transactions = []
for txn in transactions:
    # Inject Error 1: 5% of transactions fail and never hit the bank statement
    if random.random() < 0.05:
        continue
        
    bank_txn = txn.copy()
    
    # Inject Error 2: Clearance delays of 1 to 3 days
    bank_txn["Transaction_Date"] = bank_txn["Transaction_Date"] + timedelta(days=random.randint(1, 3))
    
    # Inject Error 3: Fuzzy matching discrepancies (abbreviations, missing suffixes, case changes)
    if random.random() < 0.30:
        name = bank_txn["Customer_Name"]
        name = name.replace(" Ltd", "").replace(" Limited", "").replace(" Media", "")
        name = name.replace("Reliance", "Rel").replace("Consultancy", "Consult")
        if random.random() < 0.5:
            name = name.upper() 
        bank_txn["Customer_Name"] = name
        
    # Map to completely different column names to mimic a real bank export
    bank_transactions.append({
        "Bank_Ref_ID": f"BNK-{random.randint(10000, 99999)}",
        "Clearance_Date": bank_txn["Transaction_Date"].strftime('%Y-%m-%d'),
        "Depositor_Name": bank_txn["Customer_Name"],
        "Amount_Settled": bank_txn["Amount_Billed"]
    })

# Create the Bank Payout Statement (Dataset B)
bank_df = pd.DataFrame(bank_transactions)
bank_df.to_csv("bank_payout_statement.csv", index=False)

print(f"Success: Generated internal_sales_ledger.csv with {len(ledger_df)} rows.")
print(f"Success: Generated bank_payout_statement.csv with {len(bank_df)} rows.")
```
